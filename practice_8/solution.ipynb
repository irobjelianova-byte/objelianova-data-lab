```python
# Підключаємо бібліотеки pandas та numpy
import pandas as pd
import numpy as np

# Створюємо словник з даними замовлень
data = {
    "source":  ["web", "web", "store", "app", "app", "web", "store", "web", "app", "app"],
    "amount":  [250.0, 1300.0, 90.5, np.nan, 640.0, 180.0, 415.0, np.nan, 2100.0, 320.0],
    "items":   [1, 3, 2, 3, 2, 3, 1, 1, 2, 1],
    "express": ["yes", "yes", "no", "yes", "no", "no", "no", "yes", "no", "no"],
}

# Створюємо назви замовлень
labels = ["ord_1", "ord_2", "ord_3", "ord_4", "ord_5",
          "ord_6", "ord_7", "ord_8", "ord_9", "ord_10"]
```

```python
# Завдання 1 — створення таблиці

# Завдання 1.1. Перетворюємо словник data на DataFrame
orders = pd.DataFrame(data)

# Виводимо таблицю
print(orders)
```

```python
# Завдання 1.2. Замінюємо індекси таблиці на значення зі списку labels
orders.index = labels

# Виводимо таблицю з номерами замовлень
print(orders)
```

```python
# Завдання 8 — пропущені значення

# Завдання 8.1. Виводимо замовлення, у яких amount заповнена
print(orders[orders["amount"].notna()])
```

```python
# Завдання 8.2. Виводимо замовлення, у яких amount відсутня
print(orders[orders["amount"].isna()])
```

```python
# Завдання 8.3. Створюємо копію таблиці
orders_filled= orders.copy()

# Обчислюємо медіану колонки amount
median_amount = orders_filled["amount"].median()

# Заповнюємо пропущені значення amount медіаною
orders_filled["amount"] = orders_filled["amount"].fillna(median_amount)

# Виводимо медіану
print(median_amount)

# Перевіряємо, скільки порожніх значень залишилося
print(orders_filled["amount"].isna().sum())
```

```python
# Завдання 8.4. Створюємо таблицю без рядків із порожнім amount
orders_without_na = orders.dropna(subset=["amount"])

# Виводимо розміри отриманої таблиці
print(orders_without_na.shape)
```

```python
# Завдання 9 — зміна таблиці

# Завдання 9.1. Створюємо копію таблиці
work = orders.copy()

# Збільшуємо всі значення amount на 50
work["amount"] = work["amount"] + 50

# Виводимо amount для ord_1 в оригінальній та зміненій таблицях
print(orders.loc["ord_1", "amount"])
print(work.loc["ord_1", "amount"])
```

```python
# Завдання 9.2. Додаємо колонку avg_item — середня вартість одного товару
work["avg_item"] = (work["amount"] / work["items"]).round(2)

# Виводимо таблицю з новою колонкою
print(work)
```

```python
# Завдання 9.3. Перейменовуємо колонку source на channel
work = work.rename(columns={"source": "channel"})

# Виводимо таблицю після перейменування
print(work)
```

```python
# Завдання 9.4. Видаляємо колонку express
work = work.drop(columns=["express"])

# Виводимо список колонок, що залишилися
print(work.columns)
```
