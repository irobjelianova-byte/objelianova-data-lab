```python
# Завдання 1. Рядок підключення до бази
# Вхідні рядки DSN
DSN_1 = "postgresql://etl_user:s3cr3t@db.datalab.local:5432/warehouse"
DSN_2 = "mysql://root:qwerty@localhost:3306/shop"
```

```python
#Завдання1.1–1.2. Функція для розбору DSN на окремі частини
def parse_dsn(dsn):
    # Відділяємо схему від решти рядка
    scheme, rest = dsn.split("://")
    # Відділяємо дані користувача і пароля від адреси сервера
    user_info, server_info = rest.split("@")
    # Відділяємо користувача від пароля
    user, password = user_info.split(":")
    # Відділяємо хост від порту та назви бази
    host_port, dbname = server_info.split("/")
    # Відділяємо хост від порту
    host, port = host_port.split(":")
    # Повертаємо словник із потрібними ключами
    return {
        "scheme": scheme,
        "user": user,
        "password": password,
        "host": host,
        "port": int(port),
        "dbname": dbname
    }
```

```python
#Завдання 1.3. Викликаємо функцію для обох DSN
parsed_dsn_1 = parse_dsn(DSN_1)
parsed_dsn_2 = parse_dsn(DSN_2)
# Виводимо отримані словники
print("DSN_1:", parsed_dsn_1)
print("DSN_2:", parsed_dsn_2)
```

```python
#Завдання 1.4. Функція для безпечного відображення DSN без справжнього пароля
def safe_dsn(dsn):
    # Розбираємо DSN на окремі частини
    data = parse_dsn(dsn)
    # Збираємо DSN назад, замінюючи пароль на три зірочки
    return (
        f"{data['scheme']}://"
        f"{data['user']}:***@"
        f"{data['host']}:{data['port']}/"
        f"{data['dbname']}"
    )
# Викликаємо функцію для DSN_1 та виводимо результат
print("Безпечний DSN_1:", safe_dsn(DSN_1))
```

```python
# Завдання 2.1. Функція normalize_headers
# Вхідний список назв колонок
HEADERS = ["  User ID ", "Order-Date", "Total Amount",
           "shipping address", "Customer  Name"]
# Створюємо функцію для нормалізації назв колонок
def normalize_headers(headers):
    # Створюємо новий порожній список
    normalized = []
    # Перебираємо всі назви колонок
    for header in headers:
        # Додаємо назву до нового списку
        normalized.append(header)
    # Повертаємо новий список такої самої довжини
    return normalized
# Викликаємо функцію
result = normalize_headers(HEADERS)
# Виводимо результат
print(result)
```

```python
# Завдання 2.2. Очищення та приведення назв колонок
# Створюємо функцію для обробки назв колонок
def normalize_headers(headers):
    # Створюємо порожній список для результату
    normalized = []
    # Перебираємо кожну назву колонки
    for header in headers:
        # Прибираємо пробіли на початку та в кінці
        name = header.strip()
        # Переводимо назву в нижній регістр
        name = name.lower()
        # Замінюємо пробіли на підкреслення
        name = name.replace(" ", "_")
        # Замінюємо дефіси на підкреслення
        name = name.replace("-", "_")
        # Додаємо оброблену назву до списку
        normalized.append(name)
    # Повертаємо новий список
    return normalized
# Викликаємо функцію
result = normalize_headers(HEADERS)
# Виводимо результат
print(result)
```

```python
# Завдання 2.3. Усунення подвійних підкреслень
# Створюємо функцію для повної нормалізації назв
def normalize_headers(headers):
    # Створюємо порожній список для результату
    normalized = []
    # Перебираємо кожну назву колонки
    for header in headers:
        # Прибираємо пробіли на початку та в кінці
        name = header.strip()
        # Переводимо назву в нижній регістр
        name = name.lower()
        # Замінюємо пробіли та дефіси на підкреслення
        name = name.replace(" ", "_").replace("-", "_")
        # Поки є подвійне підкреслення, замінюємо його на одне
        while "__" in name:
            name = name.replace("__", "_")
        # Додаємо готову назву до списку
        normalized.append(name)
    # Повертаємо нормалізований список
    return normalized
# Отримуємо нормалізовані назви
result = normalize_headers(HEADERS)
# Виводимо результат
print(result)
```

```python
# Завдання 2.4. Виведення назв у форматі "було -> стало"
# Отримуємо нормалізовані назви колонок
normalized_headers = normalize_headers(HEADERS)
# Перебираємо початкові та нові назви одночасно
for old, new in zip(HEADERS, normalized_headers):
    # Виводимо початкову та нормалізовану назви
    print(f"{old} -> {new}")
```

```python
# Завдання 4.1. Видалення дублікатів за id
# Вхідний список рядків
ROWS = [
    {"id": 1, "name": "Ann"},
    {"id": 2, "name": "Bob"},
    {"id": 1, "name": "Anna"},
    {"id": 3, "name": "Cid"},
    {"id": 2, "name": "Bobby"},
    {"id": 1, "name": "A."},
]
# Створюємо функцію для видалення дублікатів
def unique_by_id(rows):
    # Створюємо порожній список для унікальних рядків
    unique_rows = []
    # Створюємо множину для вже побачених id
    seen_ids = set()
    # Перебираємо всі рядки
    for row in rows:
        # Перевіряємо, чи id ще не зустрічався
        if row["id"] not in seen_ids:
            # Запам'ятовуємо id
            seen_ids.add(row["id"])
            # Додаємо рядок до результату
            unique_rows.append(row)
    # Повертаємо новий список
    return unique_rows
# Викликаємо функцію
result = unique_by_id(ROWS)
# Виводимо результат
print(result)
```

```python
# Завдання 4.2. Залишаємо перший рядок для кожного id
# Створюємо функцію для видалення дублікатів
def unique_by_id(rows):
    # Створюємо порожній список для результату
    unique_rows = []
    # Створюємо множину вже побачених id
    seen_ids = set()
    # Перебираємо рядки у початковому порядку
    for row in rows:
        # Перевіряємо, чи зустрічався цей id раніше
        if row["id"] not in seen_ids:
            # Додаємо id до множини
            seen_ids.add(row["id"])
            # Додаємо перший рядок з цим id до результату
            unique_rows.append(row)
    # Повертаємо результат у порядку першої появи
    return unique_rows
# Отримуємо список без дублікатів
result = unique_by_id(ROWS)
# Виводимо результат
for row in result:
    print(row)
```

```python
# Завдання 4.3. Перевірка, що вхідний список не змінюється
# Запам'ятовуємо початкову кількість рядків
rows_before = len(ROWS)
# Викликаємо функцію
result = unique_by_id(ROWS)
# Перевіряємо кількість рядків після виклику
rows_after = len(ROWS)
# Виводимо кількість рядків до і після
print("До:", rows_before)
print("Після:", rows_after)
```

```python
# Завдання 4.4. Підрахунок кількості рядків і дублікатів
# Отримуємо список без дублікатів
result = unique_by_id(ROWS)
# Рахуємо кількість рядків до видалення дублікатів
rows_before = len(ROWS)
# Рахуємо кількість рядків після видалення дублікатів
rows_after = len(result)
# Рахуємо кількість видалених дублікатів
duplicates = rows_before - rows_after
# Виводимо три необхідні числа
print(rows_before)
print(rows_after)
print(duplicates)
```

```python
# Завдання 5.1. Товари, які були додані
# Дані за вчора
YESTERDAY = {101: 10.0, 102: 20.0, 103: 30.0, 104: 40.0}
# Дані за сьогодні
TODAY = {102: 20.0, 103: 35.0, 105: 50.0, 106: 60.0}
# Знаходимо товари, які є сьогодні, але яких не було вчора
added = sorted(set(TODAY) - set(YESTERDAY))
# Виводимо список доданих товарів
print(added)
```

```python
# Завдання 5.2. Товари, які були видалені
# Знаходимо товари, які були вчора, але зникли сьогодні
removed = sorted(set(YESTERDAY) - set(TODAY))
# Виводимо список видалених товарів
print(removed)
```

```python
# Завдання 5.3. Товари, які залишилися
# Знаходимо товари, які присутні і вчора, і сьогодні
stayed = sorted(set(YESTERDAY) & set(TODAY))
# Виводимо список товарів, які залишилися
print(stayed)
```

```python
# Завдання 5.4. Товари, у яких змінилася ціна
# Створюємо порожній список для товарів зі зміненою ціною
changed = []
# Перевіряємо товари, які є в обох днях
for product_id in stayed:
    # Порівнюємо ціну товару за вчора і сьогодні
    if YESTERDAY[product_id] != TODAY[product_id]:
        # Додаємо товар до списку змінених
        changed.append(product_id)
# Виводимо список товарів зі зміненою ціною
print(changed)
```

```python
# Завдання 5.5. Виведення всіх чотирьох списків
# Виводимо список доданих товарів
print("added:", sorted(added))
# Виводимо список видалених товарів
print("removed:", sorted(removed))
# Виводимо список товарів, які залишилися
print("stayed:", sorted(stayed))
# Виводимо список товарів зі зміненою ціною
print("changed:", sorted(changed))
```

```python
# Завдання 6.1. Розбір логу на список словників
# Вхідний лог
LOG = """2025-03-01 02:00:01 INFO orders extract 15230 12.4
2025-03-01 02:00:14 INFO orders transform 15230 3.1
2025-03-01 02:00:18 ERROR orders load 0 0.0
2025-03-01 02:05:00 INFO orders load 15230 45.9
2025-03-01 03:00:02 INFO customers extract 812 1.2
2025-03-01 03:00:04 WARN customers transform 0 0.4
2025-03-01 03:00:05 ERROR customers load 0 0.0
2025-03-02 02:00:01 INFO orders extract 15990 13.0"""
# Створюємо порожній список для записів
records = []
# Розбиваємо лог на окремі рядки
lines = LOG.split("\n")
# Перебираємо кожен рядок
for line in lines:
    # Розкладаємо рядок на сім частин
    date, time, level, pipeline, step, rows, duration = line.split()
    # Створюємо словник із даними рядка
    record = {
        "date": date,
        "time": time,
        "level": level,
        "pipeline": pipeline,
        "step": step,
        "rows": rows,
        "duration": duration
    }
    # Додаємо словник до списку записів
    records.append(record)
# Виводимо отримані записи
print(records)
```

```python
# Завдання 6.2. Перетворення rows та duration у числові типи
# Перебираємо всі записи
for record in records:
    # Перетворюємо кількість рядків у ціле число
    record["rows"] = int(record["rows"])
    # Перетворюємо тривалість у число з плаваючою крапкою
    record["duration"] = float(record["duration"])
# Виводимо кількість зібраних записів
print(len(records))
```

```python
# Завдання 6.3. Кількість рядків за кожним рівнем
# Створюємо порожній словник для підрахунку
records_by_level = {}
# Перебираємо всі записи
for record in records:
    # Отримуємо рівень поточного запису
    level = record["level"]
     # Збільшуємо кількість записів для цього рівня на один
    records_by_level[level] = records_by_level.get(level, 0) + 1
# Виводимо результат
print(records_by_level)
```

```python
# Завдання 6.4. Сума duration за кожним pipeline
# Створюємо порожній словник для сум тривалості
duration_by_pipeline = {}
# Перебираємо всі записи
for record in records:
    # Отримуємо назву pipeline
    pipeline = record["pipeline"]
    # Додаємо тривалість до відповідного pipeline
    duration_by_pipeline[pipeline] = (
        duration_by_pipeline.get(pipeline, 0) + record["duration"]
    )
# Виводимо результат
print(duration_by_pipeline)
```

```python
# Завдання 6.5. Виведення записів з рівнем ERROR
# Перебираємо всі записи
for record in records:
    # Перевіряємо, чи рівень запису ERROR
    if record["level"] == "ERROR":
        # Виводимо дату, час, pipeline та крок
        print(
            record["date"],
            record["time"],
            record["pipeline"],
            record["step"]
        )
```

```python
# Завдання 6.6. Пошук запису з найбільшою кількістю рядків
# Знаходимо запис із максимальним значенням rows
max_record = max(records, key=lambda record: record["rows"])
# Виводимо кількість рядків, pipeline та крок
print(
    max_record["rows"],
    max_record["pipeline"],
    max_record["step"]
)
```

```python
# Завдання 7.1. Підсумки продажів за регіонами
# Вхідні дані про продажі
SALES = [
    {"date": "2025-03-01", "region": "Захід", "product": "Кава",  "amount": 120.0},
    {"date": "2025-03-01", "region": "Захід", "product": "Чай",   "amount": 80.5},
    {"date": "2025-03-01", "region": "Центр", "product": "Кава",  "amount": 200.0},
    {"date": "2025-03-02", "region": "Захід", "product": "Какао", "amount": 60.0},
    {"date": "2025-03-02", "region": "Схід",  "product": "Какао", "amount": 45.25},
    {"date": "2025-03-02", "region": "Центр", "product": "Чай",   "amount": 30.0},
    {"date": "2025-03-03", "region": "Схід",  "product": "Сік",   "amount": 15.75},
    {"date": "2025-03-03", "region": "Центр", "product": "Кава",  "amount": 90.0},
    {"date": "2025-03-03", "region": "Захід", "product": "Сік",   "amount": 25.0},
]
# Створюємо порожній словник для сум за регіонами
by_region = {}
# Перебираємо всі продажі
for sale in SALES:
    # Отримуємо назву регіону
    region = sale["region"]
    # Додаємо суму продажу до відповідного регіону
    by_region[region] = by_region.get(region, 0) + sale["amount"]
# Виводимо результат
print(by_region)
```

```python
# Завдання 7.2. Підсумки продажів за продуктами
# Створюємо порожній словник для сум за продуктами
by_product = {}
# Перебираємо всі продажі
for sale in SALES:
    # Отримуємо назву продукту
    product = sale["product"]
    # Додаємо суму продажу до відповідного продукту
    by_product[product] = by_product.get(product, 0) + sale["amount"]
# Виводимо результат
print(by_product)
```

```python
# Завдання 7.3. Загальна та середня сума продажів
# Рахуємо загальну суму всіх продажів
total = sum(sale["amount"] for sale in SALES)
# Рахуємо середню суму одного продажу
average = total / len(SALES)
# Округлюємо обидва значення до двох знаків після коми
total = round(total, 2)
average = round(average, 2)
# Виводимо результати
print("Загальна сума:", total)
print("Середня сума:", average)
```

```python
# Завдання 8.1. Перевірка одного замовлення
# Список дозволених статусів
STATUSES = ["new", "paid", "shipped"]
# Створюємо функцію для перевірки замовлення
def check_order(order):
    # Перевіряємо, чи order_id складається тільки з цифр
    if not order["order_id"].isdigit():
        return "order_id"
    # Перевіряємо, чи customer не порожній
    if order["customer"] == "":
        return "customer"
    # Перевіряємо, чи amount можна перетворити на число
    try:
        amount = float(order["amount"])
    except ValueError:
        return "amount"
    # Перевіряємо, чи amount не менше за нуль
    if amount < 0:
        return "amount"
    # Перевіряємо, чи status є у списку дозволених статусів
    if order["status"] not in STATUSES:
        return "status"
    # Якщо всі перевірки пройдено
    return None
```

```python
# Завдання 8.2. Перевірка правил у заданому порядку
# Вхідні дані замовлень
ORDERS = [
    {"order_id": "1001", "customer": "ТОВ Ромашка",  "amount": "250.00", "status": "paid"},
    {"order_id": "1002", "customer": "ФОП Іваненко", "amount": "-10.5",  "status": "new"},
    {"order_id": "abc", "customer": "ТОВ Соняшник", "amount": "99.9",   "status": "paid"},
    {"order_id": "1004", "customer": "",             "amount": "10.0",   "status": "new"},
    {"order_id": "1005", "customer": "ФОП Петренко", "amount": "77.7",   "status": "cancelled"},
    {"order_id": "1006", "customer": "ТОВ Барвінок", "amount": "",       "status": "shipped"},
    {"order_id": "1007", "customer": "ТОВ Явір",     "amount": "0",      "status": "new"},
]
# Перевіряємо кожне замовлення
for order in ORDERS:
    # Отримуємо назву першого проблемного поля
    problem = check_order(order)
    # Виводимо результат перевірки
    print(order["order_id"], "->", problem)
```

```python
# Завдання 8.3. Розподіл замовлень на valid та invalid
# Створюємо порожній список для правильних замовлень
valid = []
# Створюємо порожній список для неправильних замовлень
invalid = []
# Перебираємо всі замовлення
for order in ORDERS:
    # Перевіряємо замовлення та отримуємо проблемне поле
    problem = check_order(order)
    # Якщо проблем немає
    if problem is None:
        # Додаємо замовлення до списку valid
        valid.append(order)
    else:
        # Додаємо пару order_id та проблемне поле до invalid
        invalid.append((order["order_id"], problem))
# Виводимо обидва списки
print("valid:", valid)
print("invalid:", invalid)
```

```python
# Завдання 8.4. Кількість валідних і невалідних записів
# Виводимо кількість валідних замовлень
print("Валідних:", len(valid))
# Виводимо кількість невалідних замовлень
print("Невалідних:", len(invalid))
```

```python
# Завдання 8.5. Виведення невалідних записів
# Перебираємо всі невалідні записи
for order_id, problem in invalid:
    # Виводимо id замовлення та проблемне поле
    print(order_id, "->", problem)
```

```python
# Завдання 9.1. Виклик джерела не більше трьох разів
# Створюємо лічильник спроб
attempts_made = {"n": 0}
def unstable_source():
    """Джерело, яке падає перші два рази, а на третій віддає дані."""
    attempts_made["n"] += 1
    if attempts_made["n"] < 3:
        raise ConnectionError("з'єднання розірвано")
    return {"rows": 100}
# Виконуємо не більше трьох спроб
for attempt in range(1, 4):
    # Виводимо номер спроби
    print("Спроба:", attempt)
    # Викликаємо джерело
    try:
        result = unstable_source()
        print(result)
    except ConnectionError:
        # Ігноруємо помилку на цьому етапі
        pass
```

```python
# Завдання 9.5. Перевірка результату після трьох спроб
# Підключаємо модуль time
import time
# Створюємо лічильник спроб
attempts_made = {"n": 0}
def unstable_source():
    """Джерело, яке падає перші два рази, а на третій віддає дані."""
    attempts_made["n"] += 1
    if attempts_made["n"] < 3:
        raise ConnectionError("з'єднання розірвано")
    return {"rows": 100}
# Створюємо змінну для результату
result = None
# Виконуємо не більше трьох спроб
for attempt in range(1, 4):
    # Виводимо номер спроби
    print("Спроба:", attempt)
    try:
        # Викликаємо джерело
        result = unstable_source()
        # Виводимо результат
        print(result)
        # Виходимо з циклу після успішної спроби
        break
    except ConnectionError as error:
        # Виводимо текст помилки
        print("Помилка:", error)
        # Робимо паузу перед наступною спробою
        time.sleep(1)
# Перевіряємо, чи вдалася хоча б одна спроба
if result is None:
    # Виводимо повідомлення, якщо джерело недоступне
    print("джерело недоступне після 3 спроб")
```

```python
# Завдання 10.1. Отримання назв колонок
# Вхідні табличні дані
CSV_TEXT = """order_id,customer,amount,status
1,Ann,10.5,paid
2,Bob,7.0,new
3,Cid,3.25,paid
4,Dan,99.0,cancelled
5,Eve,1.75,paid
6,Fay,20.0,new
7,Gil,5.5,paid"""
# Розбиваємо текст на окремі рядки
lines = CSV_TEXT.split("\n")
# Беремо перший рядок і розбиваємо його по комі
columns = lines[0].split(",")
# Виводимо список назв колонок
print(columns)
```

```python
# Завдання 10.2. Створення списку словників
# Створюємо порожній список для рядків таблиці
rows = []
# Беремо всі рядки, крім першого
for line in lines[1:]:
    # Розбиваємо рядок на окремі значення
    values = line.split(",")
    # З'єднуємо назви колонок зі значеннями та створюємо словник
    row = dict(zip(columns, values))
    # Додаємо словник до списку
    rows.append(row)
# Виводимо кількість створених рядків
print(len(rows))
```

```python
# Завдання 10.3. Виведення першого словника
# Виводимо перший словник зі списку rows
print(rows[0])
```

```python
# Завдання 10.4. Перетворення amount на float
# Перебираємо всі рядки таблиці
for row in rows:
    # Перетворюємо значення amount з рядка на число
    row["amount"] = float(row["amount"])
# Виводимо перший рядок для перевірки
print(rows[0])
```

```python
# Завдання 10.5. Сума продажів зі статусом paid
# Створюємо змінну для суми
paid_total = 0
# Перебираємо всі рядки
for row in rows:
    # Перевіряємо, чи статус замовлення paid
    if row["status"] == "paid":
        # Додаємо суму замовлення
        paid_total += row["amount"]
# Округлюємо суму до двох знаків після коми
paid_total = round(paid_total, 2)
# Виводимо результат
print(paid_total)
```
