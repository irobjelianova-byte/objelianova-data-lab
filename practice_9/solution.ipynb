```python
# Завдання 7 — дати й час

# Завдання 7.1 — перевіряємо типи колонок Order Date і Ship Date
print("Тип Order Date:", orders["Order Date"].dtype)
print("Тип Ship Date:", orders["Ship Date"].dtype)
```

```python
# Завдання 7.2 — створюємо копію таблиці для роботи з датами
orders_dates =orders.copy()

# Додаємо рік замовлення
orders_dates["Order Year"] = orders_dates["Order Date"].dt.year


# Додаємо місяць замовлення
orders_dates["Order Month"] = orders_dates["Order Date"].dt.month


# Додаємо назву дня тижня
orders_dates["Order Weekday"] = orders_dates["Order Date"].dt.day_name()

# Додаємо квартал замовлення
orders_dates["Order Quarter"] = orders_dates["Order Date"].dt.quarter

print("\nКількість замовлень за кожним роком:")
print(orders_dates["Order Year"].value_counts().sort_index())
```

```python
# Завдання 7.3 — обчислюємо кількість днів між замовленням та відвантаженням
orders_dates["Ship Delay"] = (
    orders_dates["Ship Date"] - orders_dates["Order Date"]
).dt.days

# Виводимо мінімальну, максимальну та середню затримку
print("\nЗатримка відвантаження:")
print("Мінімум:", orders_dates["Ship Delay"].min(),"днів")
print("Максимум:", orders_dates["Ship Delay"].max(),"днів")
print("Середнє:",round(orders_dates["Ship Delay"].mean(),2),"днів")
```

```python
# Завдання 7.4 — створюємо помісячний ряд сумарних продажів
monthly_sales = (
    orders_dates
    .set_index("Order Date")["Sales"]
    .resample("ME")
    .sum()
)

print("\nПомісячний ряд сумарних продажів:")
print(monthly_sales)

print("\nКількість точок у помісячному ряді:", len(monthly_sales))
```

```python
# Завдання 7.5 — рахуємо середню затримку відвантаження за кожним днем тижня
average_delay_by_weekday = (
    orders_dates
    .groupby("Order Weekday")["Ship Delay"]
    .mean()
)

print("n\Середня затримка відвантаження за днями тижня:")
print(average_delay_by_weekday)
```

```python
# Завдання 8 — похідні колонки

# Створюємо копію таблиці із Завдання 7, щоб не змінювати orders_dates
orders_derived = orders_dates.copy()
```

```python
# Завдання 8.1 — одним виразом додаємо одразу три нові колонки
orders_derived = orders_derived.assign(
    Profit_Margin=orders_derived["Profit"] / orders_derived["Sales"], #маржа прибутку
    Revenue_Per_Unit=orders_derived["Sales"] / orders_derived["Order Quantity"], #продажі на одиницю
    Is_Profitable=orders_derived["Profit"] > 0 #ознака прибутковості
)

print("8.1 - Перші рядки похідних колонок:")
print(
    orders_derived[
            ["Profit_Margin", "Revenue_Per_Unit", "Is_Profitable"]
    ].head()
)
```

```python
# Завдання 8.2 — два підходи до визначення рівня знижки

# Перший підхід — створюємо функцію для визначення рівня знижки
def get_discount_level(discount):
    # Визначаємо категорію знижки за заданими умовами
    if discount == 0:
        return "No"
    elif discount < 0.05:
        return "Low"
    else:
        return "High"


# Застосовуємо функцію до кожного значення Discount
discount_by_apply = orders_derived["Discount"].apply(get_discount_level)


# Другий підхід — використовуємо pd.cut
discount_by_cut = pd.cut(
    orders_derived["Discount"],
    bins=[-float("inf"), 0, 0.05, float("inf")],
    labels=["No", "Low", "High"],
    right=False
)

# Перетворюємо результат pd.cut у звичайний текстовий тип
discount_by_cut = discount_by_cut.astype(object)


# Перевіряємо, що результати двох підходів однакові
print("\n8.2 — Чи збігаються два підходи:")
print(discount_by_apply.equals(discount_by_cut))


# Зберігаємо результат у колонку Discount Level
orders_derived["Discount Level"] = discount_by_apply

print("\nПриклади Discount Level:")
print(
    orders_derived[
        ["Discount", "Discount Level"]
    ].drop_duplicates().sort_values("Discount")
)
```

```python
# Завдання 8.3 — створюємо словник відповідності регіонів і кодів
region_codes = {
    "West": "W",
    "Atlantic": "ATL",
    "Ontario": "ON",
    "Quebec": "QC",
    "Prarie": "PRA",
    "Nunavut": "NU",
    "Northwest Territories": "NWT",
    "Yukon": "YT"
}

# Створюємо нову колонку Region Code за допомогою map
orders_derived["Region Code"] = orders_derived["Region"].map(region_codes)

# Перевіряємо кількість пропусків у Region Code
region_code_missing = orders_derived["Region Code"].isna().sum()

print("\nУнікальні коди регіонів:")
print(orders_derived["Region Code"].unique())

print("Кількість пропусків у Region Code", region_code_missing)
```

```python
# Завдання 8.4 — створюємо категорію швидкості доставки на основі Ship Delay
orders_derived["Delivery Speed"] = pd.cut(
    orders_derived["Ship Delay"],
    bins=[-float("inf"), 2, 5, float("inf")],
    labels=["Швидке", "Звичайне", "Повільне"]
)

# Рахуємо кількість записів у кожній категорії
delivery_speed_counts = orders_derived["Delivery Speed"].value_counts()

print("n\8.4 - Кількість записів за швидкістю доставки:")
print(delivery_speed_counts)
```

```python
# Завдання 9 — сортування та ранжування

# Завдання 9.1 — сортуємо таблицю за Sales у порядку спадання
orders_by_sales = orders_derived.sort_values(
    "Sales",
    ascending=False
)

# Виводимо найбільше значення продажів
print("9.1 - Найбільше значення Sales:")
print(orders_by_sales["Sales"].iloc[0])

# Сортуємо таблицю одночасно за Region за зростанням і Profit за спаданням
orders_sorted = orders_derived.sort_values(
    ["Region", "Profit"],
    ascending=[True, False]
)

print("n\Таблиця відсортована за Region і Profit:")
print(orders_sorted[["Region", "Profit"]].head(10))
```

```python
# Завдання 9.2 — знаходимо 10 найприбутковіших записів
top_10_profit = orders_derived.nlargest(
    10,
    "Profit"
)

print("\n9.2 - 10 найприбутковіших записів:")
print(top_10_profit)

# Знаходимо 10 найбільш збиткових записів
bottom_10_profit = orders_derived.nsmallest(
    10,
    "Profit"
)

print("\n10 - найбільш збиткових записів:")
print(bottom_10_profit)
```

```python
# Завдання 9.3 — створюємо ранг продажів усередині кожного регіону
ordrers_ranked = orders_derived.copy()

ordrers_ranked["Sales Rank"] = (
    ordrers_ranked
    .groupby("Region")["Sales"]
    .rank(
        ascending=False,
        method="min"
    )
)

# Відбираємо запис із рангом 1 для регіону West
west_top = ordrers_ranked[
        (ordrers_ranked["Region"] == "West") &
        (ordrers_ranked["Sales Rank"] == 1)
]

print("\n9.3 - Запис із рангом 1 для регіону West:")
print(west_top)
```

```python
# Завдання 10 — групування та агрегація


# Завдання 10.1 — рахуємо сумарні Sales і Profit за кожною Product Category
category_summary = (
    orders_derived
    .groupby("Product Category")[["Sales", "Profit"]]
    .sum()
)

print("10.1 — Сумарні Sales і Profit за категоріями:")
print(category_summary)
```

```python
# Завдання 10.2 — рахуємо чотири показники для кожного Region
region_summary = (
    orders_derived
    .groupby("Region")
    .agg(
        total_sales=("Sales", "sum"),                    # сумарні продажі
        avg_discount=("Discount", "mean"),              # середня знижка
        orders_count=("Order ID", "nunique"),            # кількість унікальних замовлень
        max_profit=("Profit", "max")                     # максимальний прибуток
    )
)

print("\n10.2 — Показники за регіонами:")
print(region_summary)
```

```python
# Завдання 10.3 — рахуємо сумарний прибуток за парою категорія + підкатегорія
category_subcategory_profit = (
    orders_derived
    .groupby(["Product Category", "Product Sub-Category"])["Profit"]
    .sum()
)

print("\n10.3 — Сумарний прибуток за категорією та підкатегорією:")
print(category_subcategory_profit)
```

```python
# Завдання 10.4 — створюємо копію таблиці для додавання похідних колонок
orders_grouped = orders_derived.copy()

# Розраховуємо середні продажі для кожної категорії без зміни кількості рядків
orders_grouped["Category Avg Sales"] = (
    orders_grouped
    .groupby("Product Category")["Sales"]
    .transform("mean")
)

# Розраховуємо відхилення кожного продажу від середнього по категорії
orders_grouped["Sales Deviation"] = (
    orders_grouped["Sales"] - orders_grouped["Category Avg Sales"]
)

print("\n10.4 — Приклади середніх продажів та відхилень:")
print(
    orders_grouped[
        [
            "Product Category",
            "Sales",
            "Category Avg Sales",
            "Sales Deviation"
        ]
    ].head(10)
)

# Перевіряємо, що кількість рядків не змінилася
print("\nКількість рядків до:", len(orders_derived))
print("Кількість рядків після:", len(orders_grouped))
```

```python
# Завдання 10.5 — знаходимо підкатегорію з найбільшим сумарним прибутком
subcategory_profit = (
    orders_derived
    .groupby(["Product Category", "Product Sub-Category"])["Profit"]
    .sum()
)

# Для кожної категорії знаходимо індекс підкатегорії з максимальним прибутком
best_subcategory = subcategory_profit.groupby(level=0).idxmax()

# Отримуємо значення прибутку для знайдених підкатегорій
best_profit = subcategory_profit.loc[best_subcategory]

# Формуємо підсумкову таблицю
best_subcategory_result = best_profit.reset_index()
best_subcategory_result.columns = [
    "Product Category",
    "Product Sub-Category",
    "Total Profit"
]

print("\n10.5 — Підкатегорія з найбільшим сумарним прибутком у кожній категорії:")
print(best_subcategory_result)
```

```python
# Завдання 11 — зведені таблиці та зміна форми


# Завдання 11.1 — будуємо зведену таблицю сумарних продажів
sales_pivot = pd.pivot_table(
    orders_derived,
    values="Sales",
    index="Region",
    columns="Product Category",
    aggfunc="sum",
    margins=True
)

print("11.1 — Зведена таблиця сумарних продажів:")
print(sales_pivot)
```

```python
# Завдання 11.2 — будуємо таблицю спряженості Customer Segment і Ship Mode
segment_ship_mode = pd.crosstab(
    orders_derived["Customer Segment"],
    orders_derived["Ship Mode"]
)

print("\n11.2 — Кількість записів за Customer Segment і Ship Mode:")
print(segment_ship_mode)
```

```python
# Завдання 11.3 — створюємо копію зведеної таблиці без загальних підсумків
sales_pivot_no_totals = pd.pivot_table(
    orders_derived,
    values="Sales",
    index="Region",
    columns="Product Category",
    aggfunc="sum"
)

# Повертаємо Region із індексу у звичайну колонку
sales_pivot_no_totals = sales_pivot_no_totals.reset_index()

# Перетворюємо широку таблицю у довгий формат
sales_long = sales_pivot_no_totals.melt(
    id_vars="Region",
    var_name="Product Category",
    value_name="Sales"
)

print("\n11.3 — Таблиця у довгому форматі:")
print(sales_long)

# Виводимо кількість рядків результату
print("\nКількість рядків у довгій таблиці:", len(sales_long))
```

```python
# Завдання 12 — об'єднання таблиць


# Завдання 12.1 — приєднуємо інформацію про повернення до всіх замовлень
orders_returns = orders_derived.merge(
    returns,
    on="Order ID",
    how="left"
)

# Заповнюємо пропуски у Status значенням No
orders_returns["Status"] = orders_returns["Status"].fillna("No")

print("12.1 — Кількість рядків після об'єднання:", len(orders_returns))

# Виводимо розподіл значень Status
print("\nРозподіл Status:")
print(orders_returns["Status"].value_counts())
```

```python
# Завдання 12.2 — перевіряємо, які Region є спільними для orders і users
common_regions = orders_derived["Region"].drop_duplicates().loc[
    lambda x: x.isin(users["Region"].unique())
]

print("\n12.2 — Регіони, які є в обох таблицях:")
print(common_regions.tolist())

# Рахуємо повторювані значення Region у таблиці users
users_region_duplicates = users["Region"].duplicated(keep=False)

print("\nКількість рядків users із повторюваними Region:")
print(users_region_duplicates.sum())

# Об'єднуємо orders з users без видалення повторів
orders_users_raw = orders_derived.merge(
    users,
    on="Region",
    how="left"
)

# Рахуємо записи, які отримали менеджера
manager_count_raw = orders_users_raw["Manager"].notna().sum()

print("\nОб'єднання users без видалення повторів:")
print("Кількість рядків:", len(orders_users_raw))
print("Кількість записів із менеджером:", manager_count_raw)


# Видаляємо дублікати Region у users, залишаючи перший запис
users_unique = users.drop_duplicates(
    subset=["Region"],
    keep="first"
)

# Повторюємо об'єднання вже з унікальним Region
orders_users_unique = orders_derived.merge(
    users_unique,
    on="Region",
    how="left"
)

# Рахуємо записи, які отримали менеджера
manager_count_unique = orders_users_unique["Manager"].notna().sum()

print("\nОб'єднання users після видалення повторів:")
print("Кількість рядків:", len(orders_users_unique))
print("Кількість записів із менеджером:", manager_count_unique)
```

```python
# Завдання 12.3 — створюємо дві маленькі таблиці для демонстрації типів об'єднання
table_a = pd.DataFrame({
    "ID": [1, 2, 3],
    "Value_A": ["A1", "A2", "A3"]
})

table_b = pd.DataFrame({
    "ID": [2, 3, 4],
    "Value_B": ["B2", "B3", "B4"]
})

# Виконуємо чотири типи об'єднання
inner_result = table_a.merge(table_b, on="ID", how="inner")
left_result = table_a.merge(table_b, on="ID", how="left")
right_result = table_a.merge(table_b, on="ID", how="right")
outer_result = table_a.merge(table_b, on="ID", how="outer")

print("\n12.3 — Кількість рядків після різних типів об'єднання:")
print("inner:", len(inner_result))
print("left:", len(left_result))
print("right:", len(right_result))
print("outer:", len(outer_result))
```

```python
# Завдання 12.4 — розділяємо таблицю на дві частини за роком замовлення
orders_before_2011 = orders_derived[
    orders_derived["Order Date"].dt.year <= 2010
]

orders_after_2010 = orders_derived[
    orders_derived["Order Date"].dt.year > 2010
]

# Склеюємо дві частини назад вертикально
orders_combined = pd.concat(
    [orders_before_2011, orders_after_2010],
    ignore_index=True
)

print("\n12.4 — Перевірка кількості рядків:")
print("Початкова кількість рядків:", len(orders_derived))
print("Кількість рядків після concat:", len(orders_combined))

# Перевіряємо, чи кількість рядків збіглася
print(
    "Кількість рядків збіглася:",
    len(orders_derived) == len(orders_combined)
)
```

```python
# Завдання 13 — бінінг, категорії та накопичувальні обчислення

# Створюємо копію таблиці для роботи
orders_13 = orders_derived.copy()


# Завдання 13.1 — розбиваємо Sales на задані діапазони
sales_bins = [0, 500, 2000, 10000, float("inf")]
sales_labels = ["Low", "Medium", "High", "Very High"]

sales_levels = pd.cut(
    orders_13["Sales"],
    bins=sales_bins,
    labels=sales_labels,
    right=False
)

print("13.1 — Кількість записів у заданих діапазонах Sales:")
print(sales_levels.value_counts().sort_index())

# Розбиваємо Sales на чотири однакові за шириною інтервали
sales_equal_width = pd.cut(
    orders_13["Sales"],
    bins=4
)

print("\nРозподіл Sales на 4 однакові за шириною інтервали:")
print(sales_equal_width.value_counts().sort_index())
```

```python
# Завдання 13.2 — розбиваємо Profit на чотири квартилі
orders_13["Profit Quartile"] = pd.qcut(
    orders_13["Profit"],
    q=4,
    labels=["Q1", "Q2", "Q3", "Q4"]
)

# Рахуємо середні продажі в кожному квартилі
avg_sales_by_quartile = (
    orders_13
    .groupby("Profit Quartile", observed=True)["Sales"]
    .mean()
)

print("\n13.2 — Середні продажі за квартилями Profit:")
print(avg_sales_by_quartile)
```

```python
# Завдання 13.3 — створюємо копії колонок для перетворення на категорійний тип
category_column = orders_13["Product Category"].copy()
region_column = orders_13["Region"].copy()

# Визначаємо використання пам'яті до перетворення
category_memory_before = category_column.memory_usage(deep=True)
region_memory_before = region_column.memory_usage(deep=True)

# Перетворюємо колонки на категорійний тип
category_column = category_column.astype("category")
region_column = region_column.astype("category")

# Визначаємо використання пам'яті після перетворення
category_memory_after = category_column.memory_usage(deep=True)
region_memory_after = region_column.memory_usage(deep=True)

print("\n13.3 — Product Category:")
print(
    "Пам'ять до:",
    round(category_memory_before / (1024 ** 2), 4),
    "MB"
)
print(
    "Пам'ять після:",
    round(category_memory_after / (1024 ** 2), 4),
    "MB"
)
print(
    "У скільки разів змінилася:",
    round(category_memory_after / category_memory_before, 4)
)

print("\n13.3 — Region:")
print(
    "Пам'ять до:",
    round(region_memory_before / (1024 ** 2), 4),
    "MB"
)
print(
    "Пам'ять після:",
    round(region_memory_after / (1024 ** 2), 4),
    "MB"
)
print(
    "У скільки разів змінилася:",
    round(region_memory_after / region_memory_before, 4)
)
```

```python
# Завдання 13.4 — сортуємо таблицю за датою замовлення
orders_cumulative = orders_13.sort_values(
    "Order Date"
).copy()

# Розраховуємо наростаючу суму продажів
orders_cumulative["Cumulative Sales"] = (
    orders_cumulative["Sales"].cumsum()
)

# Виводимо останнє значення наростаючої суми
last_cumulative_sales = orders_cumulative["Cumulative Sales"].iloc[-1]

print("\n13.4 — Останнє значення Cumulative Sales:")
print(last_cumulative_sales)

# Перевіряємо, чи дорівнює воно загальній сумі Sales
print(
    "Чи дорівнює останнє значення загальній сумі Sales:",
    last_cumulative_sales == orders_cumulative["Sales"].sum()
)
```

```python
# Завдання 13.5 — використовуємо помісячний ряд Sales із завдання 7.4
# Ковзне середнє з вікном у 3 місяці
rolling_sales = monthly_sales.rolling(
    window=3
).mean()

# Наростаюче середнє від початку періоду
expanding_sales = monthly_sales.expanding().mean()

# Абсолютна зміна між сусідніми місяцями
absolute_change = monthly_sales.diff()

# Відсоткова зміна між сусідніми місяцями
percent_change = monthly_sales.pct_change()

print("\n13.5 — Ковзне середнє за 3 місяці:")
print(rolling_sales)

print("\nНаростаюче середнє:")
print(expanding_sales)

print("\nАбсолютна зміна продажів:")
print(absolute_change)

print("\nВідсоткова зміна продажів:")
print(percent_change)

# Рахуємо кількість порожніх значень у кожному ряді
print("\nКількість порожніх значень на початку рядів:")
print("Ковзне середнє:", rolling_sales.isna().sum())
print("Наростаюче середнє:", expanding_sales.isna().sum())
print("Абсолютна зміна:", absolute_change.isna().sum())
print("Відсоткова зміна:", percent_change.isna().sum())
```

```python
# Завдання 14 — ієрархічний індекс

# Завдання 14.1 — групуємо продажі за трьома рівнями та отримуємо багаторівневий індекс
sales_hierarchical = (
    orders_derived
    .groupby(["Region", "Product Category", "Product Sub-Category"])["Sales"]
    .sum()
)

print("14.1 — Продажі за Region → Product Category → Product Sub-Category:")
print(sales_hierarchical)

print("\nКількість рівнів індексу:", sales_hierarchical.index.nlevels)
print("Кількість рядків результату:", len(sales_hierarchical))
```

```python

# Завдання 14.2 — вибираємо лише регіон West за першим рівнем індексу
west_sales = sales_hierarchical.xs("West", level="Region")

print("\n14.2 — Дані лише для регіону West:")
print(west_sales)

print("\nКількість рядків для West:", len(west_sales))

# Міняємо місцями перший і третій рівні індексу
sales_swapped = sales_hierarchical.swaplevel(
    "Region",
    "Product Sub-Category"
)

# Сортуємо результат за новим порядком індексів
sales_swapped = sales_swapped.sort_index()

print("\nПісля обміну першого і третього рівнів:")
print(sales_swapped)

print("\nНазви рівнів індексу в новому порядку:")
print(sales_swapped.index.names)
```

```python
# Завдання 14.3 — беремо результат групування з пункту 10.3
# і переносимо Product Category з рядків у колонки
category_profit = (
    orders_derived
    .groupby(["Product Category", "Product Sub-Category"])["Profit"]
    .sum()
)

category_profit_unstacked = category_profit.unstack("Product Category")

print("\n14.3 — Product Category перенесено в колонки:")
print(category_profit_unstacked)

print("\nРозмір після unstack:", category_profit_unstacked.shape)

# Повертаємо Product Category назад у рядки
category_profit_stacked = category_profit_unstacked.stack()

print("\nПісля повернення Product Category у рядки:")
print(category_profit_stacked)

print("Розмір після stack:", category_profit_stacked.shape)
```
